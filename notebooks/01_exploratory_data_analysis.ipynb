```python
import sys
from pathlib import Path

# Set repository root in sys.path
REPO_ROOT = Path.cwd().parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import config
from src.data_processing import load_raw_dataset, validate_dataset_structure, get_dataset_summary_stats
```

### 1. Data Ingestion & Structural Validation

```python
df = load_raw_dataset()
print(f"Dataset Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")

audit = validate_dataset_structure(df)
print("\n--- Audit Summary ---")
print(f"Is Valid: {audit['is_valid']}")
print(f"Missing Values: {audit['missing_values']}")
print(f"Duplicate Rows: {audit['duplicate_rows']}")
print(f"Class Distribution: {audit['class_distribution']}")
print(f"Fraud Percentage: {audit['fraud_percentage']:.4f}%")
```

### 2. Statistical Analysis: Legitimate vs Fraudulent

```python
summary = get_dataset_summary_stats(df)
legit_amounts = df[df[config.TARGET_COLUMN] == 0][config.AMOUNT_COLUMN]
fraud_amounts = df[df[config.TARGET_COLUMN] == 1][config.AMOUNT_COLUMN]

amount_comp = pd.DataFrame({
    'Legitimate': [legit_amounts.count(), legit_amounts.mean(), legit_amounts.median(), legit_amounts.std(), legit_amounts.max()],
    'Fraudulent': [fraud_amounts.count(), fraud_amounts.mean(), fraud_amounts.median(), fraud_amounts.std(), fraud_amounts.max()]
}, index=['Count', 'Mean ($)', 'Median ($)', 'Std Dev ($)', 'Max ($)'])

display(amount_comp.round(2))
```
