### Experimental Overview
This notebook documents the empirical evaluation of all four fraud detection paradigms evaluated on the held-out test partition (42,722 transactions, 74 fraud cases):
1. **Logistic Regression** (Supervised Classical Linear ML)
2. **Random Forest Classifier** (Supervised Ensemble Tree ML)
3. **Artificial Neural Network (ANN)** (Supervised Deep Learning)
4. **Autoencoder** (Unsupervised Deep Learning Anomaly Detection)

### Evaluation Metrics Focus
Because credit card fraud is highly imbalanced (0.173%), accuracy is misleadingly high (>99%). We prioritize:
- **Recall (True Positive Rate)**: Fraction of fraudulent transactions correctly flagged.
- **Precision**: Proportion of flagged transactions that were genuine fraud.
- **F1-Score**: Harmonic mean of Precision and Recall.
- **PR-AUC**: Area under the Precision-Recall Curve (primary benchmark).
